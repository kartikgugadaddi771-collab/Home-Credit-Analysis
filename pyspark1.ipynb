```python
dataset=[
 "spark is fast",
 "spark is powerful",
 "hadoop is distributed",
 "spark and hadoop are popular",
 "kafka is streaming"
]
```

```python
# Problem:
# Convert all sentences into individual words.
rdd=sc.parallelize(dataset)
result=rdd.flatMap(lambda x:x.split())
for word in result.collect():
  print(word)
```

```python
# Q19
# Problem:
# Find the total number of words.
rdd=sc.parallelize(dataset)
result=rdd.flatMap(lambda x:x.split(' '))
print(result.count())
```

```python
# Q20
# Problem:
# Find all words having more than 4 characters.
rdd=sc.parallelize(dataset)
word=rdd.flatMap(lambda x:x.split(' '))
result=word.filter(lambda w: len(w)>4)
for word in result.collect():
  print(word)
```

```python
# Q21
# Problem:
# Convert every word into uppercase.
rdd= sc.parallelize(dataset)
word=rdd.flatMap(lambda x: x.split(" "))
result=word.map(lambda w: w.upper())
for upper in result.collect():
  print(upper)
```

```python
# Q22
# Problem:
# Find the number of unique words.
rdd=sc.parallelize(dataset)
word=rdd.flatMap(lambda x:x.split(' '))
result=word.distinct().count()
print(result)
```

```python
word=[
 "spark is fast",
 "spark is powerful",
 "spark is easy",
 "hadoop is distributed",
 "spark is fast"
]
```

```python
# Q23
# Problem:
# Calculate the frequency of every word.
rdd=sc.parallelize(word)
rdd1=rdd.flatMap(lambda x:x.split(" "))
map_rdd=rdd1.groupBy(lambda word:word)
count_rdd=map_rdd.mapValues(len)
for w,c in count_rdd.collect():
  print({f"{w} {c}"})
```

```python
# Q24
# Problem:
# Find the most frequently occurring word.
# Expected Output
rdd=sc.parallelize(word)
rdd1=rdd.flatMap(lambda x: x.split(" "))
map_rdd=rdd1.map(lambda x: (x,1))
freq_count=map_rdd.reduceByKey(lambda a,b: a+b)
most_frequent=freq_count.sortBy(lambda x:x[1], ascending=False).first()
print(f"{most_frequent[0],{most_frequent[1]}}")
```

```python
# Q25
# Problem:
# Find all words that occur more than once.
rdd=sc.parallelize(word)
rdd1=rdd.flatMap(lambda x: x.split(" "))
map_rdd=rdd1.map(lambda x: (x,1))
word_counts=map_rdd.reduceByKey(lambda a,b: a+b)
result_rdd = word_counts.filter(lambda x: x[1] > 1)
for w,c in result_rdd.collect():
  print(f"{w},{c}")
```

```python
# Q26
# Problem:
# Calculate the total number of words.
rdd=sc.parallelize(word)
rdd1=rdd.flatMap(lambda x:x.split(" "))
word_count=rdd1.count()
print(word_count)
```

```python
# Q47
# Problem:
# Create an RDD containing numbers from 1 to 1000 with 20 partitions. Reduce
# the number of partitions to 5 before saving
import shutil
import os

rdd=sc.parallelize(range(1,1001),20)
print(f"Initial Partition {rdd.getNumPartitions()}")

rdd_5=rdd.coalesce(5)

print(f"After coalesce: {rdd_5.getNumPartitions()}")

# Remove output directory if it already exists
output_path = "output/numbers_1000"
if os.path.exists(output_path):
    shutil.rmtree(output_path)

rdd_5.saveAsTextFile(output_path)
```

```python
# Q50
# Problem:
# Perform the following complete analysis:
# Calculate revenue for every order.
# Keep only orders with revenue greater than 50000.
# Calculate total revenue for each product.
# Find the product with the highest revenue.
# Save the final result into a single output partition.

rdd=sc.parallelize(orders)
# Calculate revenue for every order: (order_id, product_name, order_revenue)
order_revenues=rdd.map(lambda x:(x[0],x[1],(x[2]*x[3])))

print("Revenue for every order:")
for order_id, product_name, revenue in order_revenues.collect():
  print(f"Order {order_id} ({product_name}) -> {revenue}")
print("\n")


filtered_orders_revenue=order_revenues.filter(lambda x:x[2]>50000)

print("Orders with revenue greater than 50000:")
for order_id, product_name, revenue in filtered_orders_revenue.collect():
  print(f"Order {order_id} ({product_name}) -> {revenue}")
print("\n")


product_revenue_map=rdd.map(lambda x:(x[1],x[2]*x[3]))
total_product_revenue=product_revenue_map.reduceByKey(lambda x,y:x+y)

print("Total revenue for each product:")
for product_name, total in total_product_revenue.collect():
  print(f"{product_name} -> {total}")
print("\n")


highest_revenue_product = total_product_revenue.sortBy(lambda x: x[1], ascending=False).first()
print(f"Product with highest revenue: {highest_revenue_product[0]} -> {highest_revenue_product[1]}")
print("\n")

total_product_revenue.coalesce(1).saveAsTextFile("output/final_revenue")
print("\n output:1 part file")
```
